```python
import pandas as pd
import os

# List of paths to the CSV files
csv_paths = [
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240748-751.csv',
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240751-754 .csv',
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240754-757.csv',
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240757-760 .csv',
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240760-763.csv',
    r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\240763-766.csv'
]

# Check if all files exist
for path in csv_paths:
    if not os.path.exists(path):
        print(f"File not found: {path}")
    else:
        print(f"File exists: {path}")

# Initialize an empty list to store DataFrames
dfs = []

# Read each CSV file into a DataFrame and append to the list
for csv_path in csv_paths:
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        dfs.append(df)

# Concatenate all DataFrames
combined_df = pd.concat(dfs, ignore_index=True)

# Path to save the combined CSV file
output_csv_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\sum_file\combined_data.csv'

# Save the combined DataFrame to a new CSV file
combined_df.to_csv(output_csv_path, index=False)

print(f"Combined CSV file has been saved to {output_csv_path}")
```
