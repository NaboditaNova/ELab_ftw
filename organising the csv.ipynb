```python
import pandas as pd

# Path to the CSV file
csv_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\Example Excel file_full.csv'

# Read the CSV file into a DataFrame
df = pd.read_csv(csv_path)

# Select the desired rows and columns
subset_df = df.iloc[12:22, 0:6]  # Rows 13 to 22, Columns A to F
subset_df['Ist-Menge'] = df.iloc[12:22, 8].values  # Add column I (Ist-Menge)

# Display the subset DataFrame
print("Subset of the CSV file:")
print(subset_df)
```

```python
# Rename the columns
subset_df = subset_df.rename(columns={
    'Inventar-Nr.': 'Title',
    'Benutzer/-in': 'Inventory-ID',
    'Gebindename': 'Location',
    'Füllstand': 'Purity',
    'Standort / Info': 'Manufacturer',
    'Preis / Preis pro Kg/L': 'Ist-Menge',
    'Ist-Menge': 'Opening Date'
})

# Display the updated DataFrame
print("Updated Subset of the CSV file:")
print(subset_df)
```
