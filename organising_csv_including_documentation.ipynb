##### This script is designed to read a CSV file, select and rename specific columns, add new columns, and save the modified DataFrame to a new CSV file. The script can be easily adapted to process different CSV files by changing the input file path.
 - The script starts by reading a CSV file from the specified path into a pandas DataFrame. Ensure the file path is correctly specified in the csv_path variable. **For different CSV files, change the csv_path variable to the respective file path.**

````python
import pandas as pd

# Path to the CSV file 
csv_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\20240522 .csv'

# Read the CSV file into a DataFrame
df = pd.read_csv(csv_path, encoding='latin-1')


# Display the DataFrame
print("Contents of the CSV file:")
print(df)
````

- **Column Selection:** A list of desired column names is defined
- **DataFrame Subset Creation:** A subset of the original DataFrame df is created, containing only the specified columns. This is achieved using the copy() method to ensure that the subset is an independent copy of the original DataFrame, thereby avoiding any unintentional modifications to the original data.
- **Output Display:** The subset DataFrame df_subset is printed to the console. This allows for a quick inspection of the selected columns from the large CSV file.

````python
desired_columns = ['CAS-RN', 'Substanzname', 'Detailname', 'Raum', 'Inv-No.', 'Summenformel', 'ArtikelID', 'Zusatz']
df_subset = df[desired_columns].copy()
print (df_subset)
````

- **Column Name Mapping:** A dictionary column_name_mapping is created to map the original column names to more descriptive names:

- **Column Renaming:** The columns of df_subset are renamed using the rename() method with the column_name_mapping dictionary. The inplace=True parameter ensures the changes are applied directly to df_subset.

- **Augmenting the DataFrame:** Two new columns, 'Opening date' and 'Purity', are added to df_subset, both initialized with empty strings.

- **Output Display:** The modified DataFrame df_subset is printed to the console for verification.

- **Saving the DataFrame:** The modified DataFrame df_subset is saved to a new CSV file specified by output_csv_path using the to_csv() method with index=False to exclude the DataFrame index from the output file. 
 ```output_csv_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\modified_csv_May.csv'```

- **Confirmation Message:** A confirmation message is printed to indicate that the modified DataFrame has been successfully saved to the specified file path.

````python
# Dictionary to map old column names to new column names
column_name_mapping = {
    'CAS-RN': 'CAS number',
    'Substanzname': 'Substance name',
    'Detailname': 'Title',
    'Raum': 'Location',
    'Inv-No.': 'Inventory number',
    'Summenformel': 'Structure formula',
    'ArtikelID': 'Manufacturer',
    'Zusatz': 'Batch number'
}

# Select desired columns
desired_columns = ['CAS-RN', 'Substanzname', 'Detailname', 'Raum', 'Inv-No.', 'Summenformel', 'ArtikelID', 'Zusatz']
df_subset = df[desired_columns].copy()

# Rename the columns
df_subset.rename(columns=column_name_mapping, inplace=True)
# Add new columns 'Opening date' and 'Purity'
df_subset['Opening date'] = ''
df_subset['Purity'] = ''
print(df_subset)
# Save the modified DataFrame to a new CSV file
output_csv_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\modified_csv_May.csv'
df_subset.to_csv(output_csv_path, index=False)

print(f"The modified DataFrame has been saved to {output_csv_path}")
````
